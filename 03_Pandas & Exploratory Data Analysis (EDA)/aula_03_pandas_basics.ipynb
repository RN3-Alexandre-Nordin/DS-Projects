```python
import pandas as pd
```

### Creating a simple exemple of data base

```python
data = {
    "month":["jan", "feb", "mar", "apr", "may"],
    "sales":[12000, 15000, 18000, 17000, 16000],
    "costs":[5000, 5500, 4500, 6500, 6000]
}
df =pd.DataFrame(data)
df
```

### First data reading (Basic EDA)
## EDA is very important to data scientist kwon the data and data structure
## Based on this onformation, the data scientist begins to clean and optimize the data, for this there are many different techniques for this, we will study some of them later in the course.

```python
## Print the data
df.head()
## Print data informations (data types, total of lines, total of columns and statistics)
df.info()
```

```python
## Creating new columns
## Profit column

df["profit"] = df["sales"] - df["costs"]
df
```

### Here we start to know the power of Pandas
## Get the metrics is very simple
## See the diference to calss 1 and 2

```python
df["profit"].sum()
df["profit"].mean()
df["profit"].max()
df["profit"].min()

```

```python
## Analyses with filter (analytical thinking)
## What monyhs was over the mean

df[df["profit"]>df["profit"].mean()]
```

### Exploratory Data Analysis Insights

Using Pandas made the analysis more concise and expressive.
By organizing data into a DataFrame, it became easier to compute metrics,
identify high-performing months, and compare results visually and logically.

This reinforces the importance of structured data representation in Data Science.


### Lesson exercise
## create a column "margin"
## get the major, minimal and average
## filter thwe months over de average

```python
# Creating nem margin colun
df["margin"] = df["profit"] / df["sales"] * 100
df
```

```python
# Get metrics
df["margin"].max()
df["margin"].min()
df["margin"].mean()
```

```python
# Filter the months over the average

df[df["margin"] > df["margin"].mean()]
```
