```python
import pandas as pd

# load database
df = pd.read_csv("data/titanic.csv")
```

```python
# Grouping by Class to aggregate methrics
df.groupby("Pclass")["Fare"].agg(["mean", "median", "count"])
```

```python
# Grouping by multiples fields, cross dimensions get mean of survivors
df.groupby(["Sex", "Pclass"])["Survived"].mean()
```

```python
# Ordering and ranking (comparative analyses)
df.sort_values("Fare", ascending=False).head(10)
```

```python
# Creating a rank by Fare
# Create a new field to ranking by Fare
df["Fare Ranking"] = df["Fare"].rank(ascending=False)
# Ordering by Rank
df.sort_values("Fare Ranking", ascending=False)
```

```python
# Create new field to categorize the sex field
# That is important to ML models, allways use number or binarys types in the fiels you will use in ML algorithms
df["SexCode"] = df["Sex"].map({"male": 0, "female": 1})

df[["Name", "SexCode", "Sex"]].head()
```

```python
# Vetorized operation
# Always use this instead of loops
# Creatind  new field with the Fared mean by family
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["FarePerPerson"] = df["Fare"] / df["FamilySize"]
# Print Name ordering by Family Size
df[["FamilySize", "Name"]].sort_values(by="FamilySize", ascending=False)
```

```python
# Create a category to Fare
df["FareCategory"] = df.apply(
    lambda row: "High" if row["Fare"] > df["Fare"].median() else "Low",
    axis=1
)
# Print name and Fare Category
df[["Name", "FareCategory"]].head()
# Survived by Fare Category
df.groupby("FareCategory")["Survived"].mean()
# The surve chance is bigger the the Fare was High
```

```python
# Converting to category type
df["Pclass"] = df["Pclass"].astype("category")

df[["Pclass", "Name"]].head()
# we use the Category Type to:
# Less memory
# Cear semantic
# Better reading

# Checkink Data Types
df.dtypes
```

```python
# seting Index
df.set_index("PassengerId", inplace=True)
# Index is very important to better performance

# Reset the index
df.reset_index(inplace=True)
```

```python
# Using chaining Method
(
    df
    .assign(IsHighFare=lambda x: x["Fare"] > x["Fare"].median())
    .groupby("IsHighFare")["Survived"]
    .mean()
)
# Calculate the chance to survive if you is High Fare and if not High Fare
```
