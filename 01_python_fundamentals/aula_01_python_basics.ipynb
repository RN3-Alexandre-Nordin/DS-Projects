### Data Types Explanation

- **int (integer)** represents whole numbers without decimal places.  
  It is commonly used for counting, indexing, and mathematical calculations in data analysis.

- **str (string)** represents textual data, such as names, categories, or labels.  
  Although strings are not used directly in mathematical calculations, they are essential
  for grouping, filtering, and categorizing data in Data Science.

Understanding data types is fundamental in Data Science because the type of a variable
determines which operations can be performed on it and how data is processed.

### Profit Analysis

The total profit was 50,500, with an average monthly profit of 10,100.
The highest profit value was 13,500, while the lowest was 7,000.
These results indicate a healthy and stable financial performance over the analyzed period.

```python
monthly_costs = [5000, 5500, 4500, 6500, 6000]
total_cost = sum(monthly_costs)
average_cost = total_cost / len(monthly_costs)
max_cost = max(monthly_costs)
min_cost = min(monthly_costs)

total_cost, average_cost, max_cost, min_cost

```

```python
monthly_profit = [sale - cost for sale, cost in zip(monthly_sales, monthly_costs)]

total_profit = sum(monthly_profit)
average_profit = total_profit / len(monthly_profit)
max_profit = max(monthly_profit)
min_profit = min(monthly_profit)

print(f"The total Profit is: {total_profit}")
print(f"The profit average is: {average_profit}")
print(f"The max of Profit is: {max_profit}")
print(f"The min Profit is: {min_profit}")
```
