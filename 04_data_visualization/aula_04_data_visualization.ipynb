```python
import pandas as pd
# load database
df = pd.read_csv("data/titanic.csv")
```

```python
# Imporeting the necessary Library
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set the Style
sns.set(style="whitegrid")
```

```python
# Plot the survival Rate by Sex
# mean by Sex, Plot the percentage of Survived, because Survived is a boolean type
sns.barplot(
    data=df,
    x="Sex",
    y="Survived"
)
plt.title("Survival Rate by Sex")
plt.show
```

```python
# Counting By Sex
ax = sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)
ax.legend(
    title="Survived",
    labels=["No", "Yes"]
)
plt.title("Passenger Count By Sex")
plt.show()
```

```python
# Passenger Distribution Age
sns.histplot(
    df['Age'], bins=30, kde=True
)
plt.title("Age Distribution")
plt.show()
```

```python
# Box Plot by Fare
sns.boxplot(data=df,
            x="Pclass",
            y="Fare")
plt.title("Fare Distribution by Class")
plt.show()
```

### Visualization Insights

Visualizations helped confirm patterns identified during EDA.
Survival rates varied significantly by gender, class, and fare category,
reinforcing the importance of socioeconomic factors.

Well-chosen plots transformed numerical summaries into intuitive evidence.
