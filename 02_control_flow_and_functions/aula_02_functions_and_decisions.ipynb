### Lesson 2 - Automated Analysis and Functions

In this lesson, I tranformed manual data analysis intop a resuable and documented functions.
By separeting calculation logic from decision rules, the analysis became a more reliable, scalable, and aligned with a real world Data Science workflows

```python
def basic_metrics(values):
        """
        Calculate basic statistical metrics for a list of numeric values.

        Parameters:
                values(list): A list of numeric values.

        returns
                dcit: A dictionary containing total, average, max end min values.
        """
        return{
                "total": sum(values),
                "averege": sum(values) / len(values),
                "max": max(values),
                "min": min(values)
        }
```

```python
monthly_sales = [12000, 25000, 15000, 13500, 14200, 17250]
monthly_costs = [6000, 13000, 7000, 6250, 8310, 7460]
sales_metrics = basic_metrics(monthly_sales)
costs_metrics = basic_metrics(monthly_costs)

sales_metrics, costs_metrics
```

```python
def profit_analysis(sales, costs):
        """
        computes monthly profit and aggregated profit metrics

        Parameters:
            sales (list): Monthly sales values
            costs (list): Monthly costs values
        Retruns:
            dict: Dictionary with monthly profit list and profit metrics

        """

        if len(sales) != len(costs):
            raise ValueError("Sales and costs must the have the same length")
        monthly_profit = [s - c for s, c in zip(sales, costs)]

        return {
            "monthly_profit": monthly_profit,
            "metrics": basic_metrics(monthly_profit)
        }

```

```python
profit_report = profit_analysis(monthly_sales, monthly_costs)
profit_report
```

```python
def profit_status(profit_metrics, threshold):
    """
    Evaluates profit performance based on average profit

    Parameters:
        profit_metrics (dict): Dictionary whit profit metrics
        thershopld {float}: Minimumaverage profit therhold

    Returns    
        str: Performance status message
    """
    avg_profit = profit_metrics["averege"]

    if avg_profit >= threshold:
        return "Strong Performance"
    else:
        return "Needs Improvement"
```

```python
status = profit_status(profit_report["metrics"], threshold=9000)
status
```

```python
profit_report = profit_analysis(monthly_sales, monthly_costs)
status = profit_status(profit_report["metrics"], threshold=9000)

profit_report, status
```
